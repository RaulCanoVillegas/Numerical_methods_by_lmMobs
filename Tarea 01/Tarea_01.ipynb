```python
import numpy as np
from matrices import MATRICES
from solvers import chole, resolver_chol, NoDefinidaPositiva

def primer_fallo(nombre, nmax=40):
    f = MATRICES[nombre]
    for n in range(2, nmax + 1):
        try:
            chole(np.array(f(n), dtype=float))
        except NoDefinidaPositiva as e:
            return n, str(e)
    return None, "sin fallo hasta n = %d" % nmax
```

```python
for nombre in ("hilb", "cauchy"):
    n, msg = primer_fallo(nombre)
    print(nombre, "falla desde n =", n, "|", msg)
    A = MATRICES[nombre](n - 1)
    x = resolver_chol(A, A @ np.ones(n - 1))
    print("   en n =", n - 1, "| error rel:", np.linalg.norm(x - 1) / np.sqrt(n - 1),
          "| cond:", np.linalg.cond(A))
```
